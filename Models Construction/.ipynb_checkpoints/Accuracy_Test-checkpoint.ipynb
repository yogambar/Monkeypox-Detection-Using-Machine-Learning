```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import ResNet50, EfficientNetB0
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Ensure reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# Dataset directories
train_dir = 'D:\\PROJECT\\Monkeypox Detection Using Machine Learning\\DataSet\\Train'
val_dir = 'D:\\PROJECT\\Monkeypox Detection Using Machine Learning\\DataSet\\Val'

# Data augmentation (no training augmentation)
img_height, img_width = 224, 224
batch_size = 32

datagen_val = ImageDataGenerator(rescale=1./255)

val_data = datagen_val.flow_from_directory(
    val_dir, target_size=(img_height, img_width), batch_size=batch_size, class_mode='binary', shuffle=False
)

# Function to evaluate models (only accuracy)
def evaluate_model(model, model_name):
    print(f"\nEvaluating {model_name}...\n")
    val_loss, val_accuracy = model.evaluate(val_data)
    print(f"{model_name} Validation Accuracy: {val_accuracy * 100:.2f}%")
    return val_accuracy

# CNN Model (assumes it's already trained)
cnn_model = tf.keras.models.load_model('D:\\PROJECT\\Monkeypox Detection Using Machine Learning\\MODELS\\CNN.keras')
cnn_accuracy = evaluate_model(cnn_model, 'CNN')

# ResNet50 Model (assumes it's already trained)
resnet_model = tf.keras.models.load_model('D:\\PROJECT\\Monkeypox Detection Using Machine Learning\\MODELS\\ResNet50.keras')
resnet_accuracy = evaluate_model(resnet_model, 'ResNet50')

# EfficientNetB0 Model (assumes it's already trained)
efficientnet_model = tf.keras.models.load_model('D:\\PROJECT\\Monkeypox Detection Using Machine Learning\\MODELS\\EfficientNetB0.keras')
efficientnet_accuracy = evaluate_model(efficientnet_model, 'EfficientNetB0')

# Print Results
print("\nModel Accuracies:")
print(f"CNN Accuracy: {cnn_accuracy * 100:.2f}%")
print(f"ResNet50 Accuracy: {resnet_accuracy * 100:.2f}%")
print(f"EfficientNetB0 Accuracy: {efficientnet_accuracy * 100:.2f}%")
```
